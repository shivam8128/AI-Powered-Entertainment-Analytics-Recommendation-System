# Exploratory Data Analysis: Entertainment Analytics
This notebook performs EDA on the generated synthetic datasets.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='darkgrid')

# Load datasets
movies = pd.read_csv('../data/movies.csv')
ratings = pd.read_csv('../data/ratings.csv')
reviews = pd.read_csv('../data/reviews.csv')
engagement = pd.read_csv('../data/engagement.csv')
```

## 1. Movies Distribution by Genre

```python
plt.figure(figsize=(10, 6))
sns.countplot(y='genre', data=movies, order=movies['genre'].value_counts().index, palette='viridis')
plt.title('Movie Distribution by Genre')
plt.xlabel('Count')
plt.ylabel('Genre')
plt.show()
```

## 2. Ratings Heatmap / Analysis

```python
plt.figure(figsize=(8, 5))
sns.histplot(ratings['rating'], bins=5, kde=False, color='coral')
plt.title('Distribution of Movie Ratings')
plt.xlabel('Rating (1-5)')
plt.ylabel('Frequency')
plt.show()
```

## 3. Detecting Fake Engagement Anomalies

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x='views', y='likes', hue='is_anomaly', data=engagement, palette='coolwarm')
plt.title('Views vs Likes: Normal vs Fake Engagement')
plt.show()
```
